> *"Como aumentar o investimento em campanhas digitais mantendo ou reduzindo o CPA?"*

## Contexto do Negócio

### Sobre a Empresa
A **Social Wave** é uma empresa de marketing digital que investe em anúncios pagos nas principais plataformas digitais:
- **Meta Ads** (Facebook/Instagram)
- **Google Search**
- **YouTube Ads**
- **Twitter Ads**
- **TikTok**
- **LinkedIn Ads**

O objetivo é gerar **conversões** (leads, vendas, cadastros) para seus clientes.

---

### O Problema

No último trimestre 2023, a Social Wave decidiu **aumentar o orçamento de marketing**. A estratégia adotada foi **dividir o novo investimento igualmente entre todos os canais**.

**Resultado:** O **CPA (Custo Por Aquisição) disparou**.

| Antes do aumento | Após o aumento |
|------------------|----------------|
| Orçamento menor | Orçamento maior |
| CPA controlado | CPA elevado |
| Eficiência saudável | Eficiência comprometida |

**O investimento cresceu, mas a eficiência caiu.**

---

### O Desafio

**Como aumentar o investimento total nas campanhas de forma que o CPA não aumente — ou até reduza?**

A resposta não está em "gastar mais", mas em **"gastar melhor"**: alocar cada dólar no canal que oferece o melhor retorno naquele momento.

---

### O Dado Disponível

Base histórica do **último trimestre de 2023 (outubro a dezembro)**:
- Período **ANTES** do aumento de verba
- Dados granulares por **anúncio, canal e timestamp**
- Permite entender o comportamento de cada canal em seu estado "natural"

*Este dado é a chave: ele mostra como cada canal performa antes da distorção causada pela alocação igualitária.*

## Objetivo do Projeto

**Aumentar o investimento total nas campanhas de marketing mantendo ou reduzindo o CPA global.**

---

### KPI Principal

| KPI | Fórmula | O que mede | Meta |
|-----|---------|-----------|------|
| **CPA** | `Gasto ÷ Conversões` | Custo para adquirir 1 conversão | **Manter ou reduzir** ao aumentar investimento |

---

### Métricas Secundárias (suporte à análise)

| Métrica | Fórmula | O que indica | Uso neste projeto |
|---------|---------|------------|-----------------|
| **CTR** (Click-Through Rate) | `Cliques ÷ Impressões` | Atratividade do anúncio | Identificar canais com bom criativo |
| **Taxa de Conversão** | `Conversões ÷ Cliques` | Eficiência do funil | Detectar onde o funil "vaza" |
| **CPC** (Custo por Clique) | `Gasto ÷ Cliques` | Custo do tráfego | Comparar custo de atração |
| **CPM** (Custo por Mil) | `(Gasto ÷ Impressões) × 1000` | Custo de alcance | Avaliar eficiência de veiculação |
| **Share de Gasto** | `Gasto do Canal ÷ Gasto Total` | Peso do investimento | Verificar desbalanceamento |
| **Share de Conversão** | `Conversões do Canal ÷ Conversões Totais` | Peso do resultado | Cruzar com share de gasto |

---

### Trade-off Central


> **Pergunta que este projeto responde:** *Até onde cada canal escala bem, e onde começa a perder eficiência?*

### Sucesso do Projeto

O projeto será bem-sucedido se entregarmos:
