> *"Como aumentar o investimento em campanhas digitais mantendo ou reduzindo o CPA?"*

## Contexto do Negócio

### Sobre a Empresa
A **Social Wave** é uma empresa de marketing digital que investe em anúncios pagos nas principais plataformas digitais:
- **Meta Ads** (Facebook/Instagram)
- **Google Search**
- **YouTube Ads**
- **Twitter Ads**
- **TikTok**
- **LinkedIn Ads**

O objetivo é gerar **conversões** (leads, vendas, cadastros) para seus clientes.

---

### O Problema

No último trimestre 2023, a Social Wave decidiu **aumentar o orçamento de marketing**. A estratégia adotada foi **dividir o novo investimento igualmente entre todos os canais**.

**Resultado:** O **CPA (Custo Por Aquisição) disparou**.

| Antes do aumento | Após o aumento |
|------------------|----------------|
| Orçamento menor | Orçamento maior |
| CPA controlado | CPA elevado |
| Eficiência saudável | Eficiência comprometida |

**O investimento cresceu, mas a eficiência caiu.**

---

### O Desafio

**Como aumentar o investimento total nas campanhas de forma que o CPA não aumente — ou até reduza?**

A resposta não está em "gastar mais", mas em **"gastar melhor"**: alocar cada dólar no canal que oferece o melhor retorno naquele momento.

---

### O Dado Disponível

Base histórica do **último trimestre de 2023 (outubro a dezembro)**:
- Período **ANTES** do aumento de verba
- Dados granulares por **anúncio, canal e timestamp**
- Permite entender o comportamento de cada canal em seu estado "natural"

*Este dado é a chave: ele mostra como cada canal performa antes da distorção causada pela alocação igualitária.*

## Objetivo do Projeto

**Aumentar o investimento total nas campanhas de marketing mantendo ou reduzindo o CPA global.**

---

### KPI Principal

| KPI | Fórmula | O que mede | Meta |
|-----|---------|-----------|------|
| **CPA** | `Gasto ÷ Conversões` | Custo para adquirir 1 conversão | **Manter ou reduzir** ao aumentar investimento |

---

### Métricas Secundárias (suporte à análise)

| Métrica | Fórmula | O que indica | Uso neste projeto |
|---------|---------|------------|-----------------|
| **CTR** (Click-Through Rate) | `Cliques ÷ Impressões` | Atratividade do anúncio | Identificar canais com bom criativo |
| **Taxa de Conversão** | `Conversões ÷ Cliques` | Eficiência do funil | Detectar onde o funil "vaza" |
| **CPC** (Custo por Clique) | `Gasto ÷ Cliques` | Custo do tráfego | Comparar custo de atração |
| **CPM** (Custo por Mil) | `(Gasto ÷ Impressões) × 1000` | Custo de alcance | Avaliar eficiência de veiculação |
| **Share de Gasto** | `Gasto do Canal ÷ Gasto Total` | Peso do investimento | Verificar desbalanceamento |
| **Share de Conversão** | `Conversões do Canal ÷ Conversões Totais` | Peso do resultado | Cruzar com share de gasto |

# Ajustes e Validação Estrutural dos dados

```python
# Renomear coluna timestamp para Data da Coleta
df.rename(columns={'timestamp':'Data da Coleta'}, inplace=True)
```

```python
# ============================================================
# TIPOS DE DADOS
# ============================================================
print("\n" + "-" * 60)
print("TIPOS DE DADOS")
print("-" * 60)

# Verificar tipos atuais
print("\nTipos de dados por coluna:")
print(df.dtypes)

# Converter Data da Coleta para datetime
if df['Data da Coleta'].dtype != 'datetime64[ns]':
    df['Data da Coleta'] = pd.to_datetime(df['Data da Coleta'])
    print("\n✅ 'Data da Coleta' convertido para datetime")

# Verificar tipos esperados vs. reais
tipos_esperados = {
    'ID': 'object',           # string
    'Canal': 'object',        # string categórica
    'Data da Coleta': 'datetime64[ns]',
    'Impressoes Disponiveis': 'int64',
    'Consultas Correspondentes': 'int64',
    'Impressoes': 'int64',
    'Cliques': 'int64',
    'Gasto': 'float64',
    'Conversao': 'int64'
}

print("\nVerificação de tipos esperados:")
for col, tipo_esperado in tipos_esperados.items():
    tipo_real = str(df[col].dtype)
    status = "✅" if tipo_real == tipo_esperado else "⚠️"
    print(f"  {status} {col}: {tipo_real} (esperado: {tipo_esperado})")
```

```python
# ============================================================
# VALORES AUSENTES (MISSING VALUES)
# ============================================================
print("\n" + "-" * 60)
print("VALORES AUSENTES")
print("-" * 60)

missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'Coluna': missing.index,
    'Ausentes': missing.values,
    '% do Total': missing_pct.values
})

print("\n Resumo de valores ausentes:")
print(missing_df.to_string(index=False))

if missing.sum() == 0:
    print("\n✅ NENHUM valor ausente encontrado — base completa!")
else:
    print(f"\n⚠️ ATENÇÃO: {missing.sum()} valores ausentes detectados")
```

```python
# ============================================================
# REGISTROS DUPLICADOS
# ============================================================
print("\n" + "-" * 60)
print("REGISTROS DUPLICADOS")
print("-" * 60)

duplicados = df.duplicated().sum()
print(f"\nRegistros duplicados: {duplicados}")

if duplicados > 0:
    print(f"⚠️ {duplicados} registros duplicados encontrados")
    print("   Recomendação: investigar e decidir manter/remover")
else:
    print("✅ Nenhum registro duplicado encontrado")

# Verificar duplicatas no ID (chave única)
id_duplicados = df['ID'].duplicated().sum()
print(f"\nIDs duplicados: {id_duplicados}")
if id_duplicados > 0:
    print(f"⚠️ {id_duplicados} IDs repetidos — verificar integridade")
else:
    print("✅ Todos os IDs são únicos")
```

```python
# ============================================================
# PERÍODO DOS DADOS
# ============================================================
print("\n" + "-" * 60)
print("PERÍODO DOS DADOS")
print("-" * 60)

data_inicio = df['Data da Coleta'].min()
data_fim = df['Data da Coleta'].max()
dias_totais = (data_fim - data_inicio).days

print(f"\nInício: {data_inicio}")
print(f"Fim: {data_fim}")
print(f"Duração: {dias_totais} dias")
print(f"Meses cobertos: {sorted(df['Data da Coleta'].dt.month.unique())}")

# Verificar se cobre o trimestre esperado (Out-Dez 2023)
meses_esperados = [10, 11, 12]
meses_presentes = sorted(df['Data da Coleta'].dt.month.unique())
if meses_presentes == meses_esperados:
    print("✅ Período confirma Outubro a Dezembro de 2023")
else:
    print(f"⚠️ Meses presentes: {meses_presentes} (esperado: {meses_esperados})")
```

```python
# ============================================================
# CANAIS PRESENTES
# ============================================================
print("\n" + "-" * 60)
print("CANAIS PRESENTES")
print("-" * 60)

canais = df['Canal'].value_counts().sort_index()
print(f"\nTotal de canais distintos: {df['Canal'].nunique()}")

# Mapeamento de canais para plataformas
mapeamento_canais = {
    'Canal1': 'Meta Ads',
    'Canal2': 'Google Search',
    'Canal3': 'YouTube Ads',
    'Canal4': 'Twitter Ads',
    'Canal5': 'TikTok',
    'Canal6': 'LinkedIn Ads'
}

print("\nMapeamento Canal → Plataforma:\n")
for canal, plataforma in mapeamento_canais.items():
    count = canais.get(canal, 0)
    print(f"  {canal}: {plataforma} ({count:,} registros)")

# Verificar se todos os 6 canais estão presentes
canais_faltantes = set(mapeamento_canais.keys()) - set(df['Canal'].unique())
if canais_faltantes:
    print(f"\n⚠️ Canais faltantes: {canais_faltantes}")
else:
    print("\n✅ Todos os 6 canais estão presentes na base")
```

```python
# ============================================================
# CONSISTÊNCIA LÓGICA DOS DADOS
# ============================================================
print("\n" + "-" * 60)
print("CONSISTÊNCIA LÓGICA")
print("-" * 60)

# Regra 1: Impressões ≤ Impressões Disponíveis
impressoes_invalidas = (df['Impressoes'] > df['Impressoes Disponiveis']).sum()
print(f"\nImpressões > Impressões Disponíveis: {impressoes_invalidas}")
if impressoes_invalidas > 0:
    print(f"   ⚠️ {impressoes_invalidas} registros com inconsistência")
else:
    print("   ✅ Todas as impressões estão dentro do inventário")

# Regra 2: Cliques ≤ Impressões
cliques_invalidos = (df['Cliques'] > df['Impressoes']).sum()
print(f"\nCliques > Impressões: {cliques_invalidos}")
if cliques_invalidos > 0:
    print(f"   ⚠️ {cliques_invalidos} registros com mais cliques que impressões")
else:
    print("   ✅ Cliques não excedem impressões")

# Regra 3: Conversões ≤ Cliques
conversoes_invalidas = (df['Conversao'] > df['Cliques']).sum()
print(f"\nConversões > Cliques: {conversoes_invalidas}")
if conversoes_invalidas > 0:
    print(f"   ⚠️ {conversoes_invalidas} registros com mais conversões que cliques")
else:
    print("   ✅ Conversões não excedem cliques")

# Regra 4: Gasto ≥ 0
gasto_negativo = (df['Gasto'] < 0).sum()
print(f"\nGasto negativo: {gasto_negativo}")
if gasto_negativo > 0:
    print(f"   ⚠️ {gasto_negativo} registros com gasto negativo")
else:
    print("   ✅ Nenhum gasto negativo")

# Regra 5: Gasto = 0 mas Conversão > 0 (impossível)
gasto_zero_conversao = ((df['Gasto'] == 0) & (df['Conversao'] > 0)).sum()
print(f"\nGasto = 0 mas Conversão > 0: {gasto_zero_conversao}")
if gasto_zero_conversao > 0:
    print(f"   ⚠️ {gasto_zero_conversao} registros inconsistentes")
else:
    print("   ✅ Nenhuma conversão sem gasto associado")
```

```python
# ============================================================
# ESTATÍSTICAS DESCRITIVAS RÁPIDAS
# ============================================================
print("\n" + "-" * 60)
print("ESTATÍSTICAS DESCRITIVAS RÁPIDAS")
print("-" * 60)

# Colunas numéricas principais
cols_numericas = ['Impressoes Disponiveis', 'Consultas Correspondentes', 
                  'Impressoes', 'Cliques', 'Gasto', 'Conversao']

print("\nResumo estatístico:")
print(df[cols_numericas].describe().round(2).to_string())
```

```python
# ============================================================
# RESUMO DA VALIDAÇÃO
# ============================================================
print("\n" + "=" * 60)
print("RESUMO DA VALIDAÇÃO")
print("=" * 60)

# Contar problemas
problemas = []
if missing.sum() > 0:
    problemas.append(f"{missing.sum()} valores ausentes")
if duplicados > 0:
    problemas.append(f"{duplicados} registros duplicados")
if impressoes_invalidas > 0:
    problemas.append(f"{impressoes_invalidas} impressões > disponíveis")
if cliques_invalidos > 0:
    problemas.append(f"{cliques_invalidos} cliques > impressões")
if conversoes_invalidas > 0:
    problemas.append(f"{conversoes_invalidas} conversões > cliques")
if gasto_negativo > 0:
    problemas.append(f"{gasto_negativo} gastos negativos")
if gasto_zero_conversao > 0:
    problemas.append(f"{gasto_zero_conversao} conversões sem gasto")

if not problemas:
    print("\n✅ BASE VALIDADA COM SUCESSO!")
    print("   • Sem valores ausentes")
    print("   • Sem registros duplicados")
    print("   • Sem inconsistências lógicas")
    print("   • Período correto: Out-Dez 2023")
    print("   • 6 canais presentes")
    print("\n🚀 Base pronta para análise exploratória!")
else:
    print(f"\n⚠️ {len(problemas)} PROBLEMA(S) ENCONTRADO(S):")
    for p in problemas:
        print(f"   • {p}")
    print("\n🔧 Recomendação: tratar problemas antes de prosseguir")

print("=" * 60)
```

```python
print("=" * 60)
print("INSPEÇÃO DE REGISTROS INCONSISTENTES")
print("=" * 60)

# ============================================================
# 1. REGISTROS COM CLIQUES > IMPRESSÕES
# ============================================================
print("\n" + "-" * 60)
print("1. REGISTROS COM CLIQUES > IMPRESSÕES")
print("-" * 60)

# Criar máscara booleana (True onde cliques > impressões)
mascara_cliques_invalidos = df['Cliques'] > df['Impressoes']

# Contar quantos são
total_invalidos = mascara_cliques_invalidos.sum()
print(f"\nTotal de registros: {total_invalidos}")

if total_invalidos > 0:
    # Selecionar apenas os registros problemáticos
    df_cliques_invalidos = df[mascara_cliques_invalidos].copy()
    
    print(f"\nRegistros completos:")
    # Mostrar colunas relevantes ordenadas pelo "excesso" de cliques
    df_cliques_invalidos['Diferenca'] = df_cliques_invalidos['Cliques'] - df_cliques_invalidos['Impressoes']
    df_cliques_invalidos = df_cliques_invalidos.sort_values('Diferenca', ascending=False)
    
    display(df_cliques_invalidos[['ID', 'Canal', 'Data da Coleta', 'Impressoes', 'Cliques', 'Diferenca', 'Gasto', 'Conversao']].head(10))
    
    # Estatísticas dos casos problemáticos
    print(f"\nEstatísticas dos registros inválidos:")
    print(f"   • Média de Impressões: {df_cliques_invalidos['Impressoes'].mean():.2f}")
    print(f"   • Média de Cliques: {df_cliques_invalidos['Cliques'].mean():.2f}")
    print(f"   • Maior diferença (Cliques - Impressões): {df_cliques_invalidos['Diferenca'].max()}")
    print(f"   • Canais afetados: {df_cliques_invalidos['Canal'].unique()}")
    
    # Verificar se há padrão (ex: sempre o mesmo canal, mesmo horário)
    print(f"\nDistribuição por canal:")
    print(df_cliques_invalidos['Canal'].value_counts().to_string())
    
else:
    print("✅ Nenhum registro com cliques > impressões")
```

```python
# ============================================================
# 2. REGISTROS COM CONVERSÕES > CLIQUES
# ============================================================
print("\n" + "-" * 60)
print("2. REGISTROS COM CONVERSÕES > CLIQUES")
print("-" * 60)

mascara_conv_invalidas = df['Conversao'] > df['Cliques']
total_conv_invalidas = mascara_conv_invalidas.sum()
print(f"\nTotal de registros: {total_conv_invalidas}")

if total_conv_invalidas > 0:
    df_conv_invalidas = df[mascara_conv_invalidas].copy()
    df_conv_invalidas['Diferenca'] = df_conv_invalidas['Conversao'] - df_conv_invalidas['Cliques']
    df_conv_invalidas = df_conv_invalidas.sort_values('Diferenca', ascending=False)
    
    display(df_conv_invalidas[['ID', 'Canal', 'Data da Coleta', 'Cliques', 'Conversao', 'Diferenca', 'Gasto']].head(10))
    
    print(f"\nDistribuição por canal:")
    print(df_conv_invalidas['Canal'].value_counts().to_string())
else:
    print("✅ Nenhum registro com conversões > cliques")
```

```python
# ============================================================
# 3. REGISTROS COM IMPRESSÕES > IMPRESSÕES DISPONÍVEIS
# ============================================================
print("\n" + "-" * 60)
print("3. REGISTROS COM IMPRESSÕES > IMPRESSÕES DISPONÍVEIS")
print("-" * 60)

mascara_imp_invalidas = df['Impressoes'] > df['Impressoes Disponiveis']
total_imp_invalidas = mascara_imp_invalidas.sum()
print(f"\nTotal de registros: {total_imp_invalidas}")

if total_imp_invalidas > 0:
    df_imp_invalidas = df[mascara_imp_invalidas].copy()
    df_imp_invalidas['Diferenca'] = df_imp_invalidas['Impressoes'] - df_imp_invalidas['Impressoes Disponiveis']
    df_imp_invalidas = df_imp_invalidas.sort_values('Diferenca', ascending=False)
    
    display(df_imp_invalidas[['ID', 'Canal', 'Data da Coleta', 'Impressoes Disponiveis', 'Impressoes', 'Diferenca']].head(10))
else:
    print("✅ Nenhum registro com impressões > disponíveis")
```

```python
# ============================================================
# 2. APLICAR O FILTRO (REMOVER)
# ============================================================
print("\n" + "-" * 60)
print("2. APLICANDO FILTRO DE CONSISTÊNCIA")
print("-" * 60)

# Guardar dimensão original
registros_antes = len(df)
print(f"\n📊 Registros ANTES: {registros_antes:,}")

# Remover registros onde Cliques > Impressões
df = df[~mascara_cliques_invalidos].copy()

# Guardar dimensão nova
registros_depois = len(df)
print(f"📊 Registros DEPOIS: {registros_depois:,}")
print(f"📊 Removidos: {registros_antes - registros_depois}")
print(f"📊 % da base removida: {((registros_antes - registros_depois) / registros_antes * 100):.2f}%")
```

```python
# ============================================================
# 3. VALIDAR QUE A INCONSISTÊNCIA FOI ELIMINADA
# ============================================================
print("\n" + "-" * 60)
print("3. VALIDAÇÃO PÓS-TRATAMENTO")
print("-" * 60)

# Re-verificar a mesma condição
cliques_restantes_invalidos = (df['Cliques'] > df['Impressoes']).sum()
print(f"\n✅ Registros com Cliques > Impressões restantes: {cliques_restantes_invalidos}")

# Verificar outras consistências também
print(f"✅ Impressões > Disponíveis: {(df['Impressoes'] > df['Impressoes Disponiveis']).sum()}")
print(f"✅ Conversões > Cliques: {(df['Conversao'] > df['Cliques']).sum()}")
print(f"✅ Gasto negativo: {(df['Gasto'] < 0).sum()}")
```
